## Types of prompts

You may know this already - but if not, you will get very familiar with it!

Models like Gemini have been trained to receive instructions in a particular way.

They expect to receive:

**A system prompt** that tells them what task they are performing and what tone they should use

**A user prompt** -- the conversation starter that they should reply to

````python
# Define our system prompt - you can experiment with this later, changing the last sentence to 'Respond in markdown in Spanish."

system_prompt = """
You are a snarky assistant that analyzes the contents of a website,
and provides a short, snarky, humorous summary, ignoring text that might be navigation related.
Respond in markdown. Do not wrap the markdown in a code block - respond just with the markdown.
"""
````

````python
# Define our user prompt

user_prompt_prefix = """
Here are the contents of a website.
Provide a short summary of this website.
If it includes news or announcements, then summarize these too.

"""


#  System prompt is the personality/instructions, user prompt is the specific request. This pattern is used everywhere in LLM engineering.
# The system prompt stays the same. It's like setting up the AI's personality once at the start.
# But: Each time you call the API, you send a fresh list of messages. It's not one continuous chat that remembers everything.
````

## Messages

The API from Gemini expects to receive messages in a structure like:

```python
[
    {"role": "user", "parts": ["user message goes here"]}
]
```
Or simply passing the text directly. System instructions are passed separately in the configuration. To give you a preview, the next 2 cells make a rather simple call.


````python
from google import genai
import os

client = genai.Client()  # Initialize the Gemini client using the modern google-genai SDK
 
messages = [
    {"role": "user", "content": "What is 2 + 2?"}
]  # Define a simple messages list containing a role and content dictionary
 
response = client.models.generate_content(
    model="gemini-2.5-flash",  # Specify the lightweight and fast Gemini model
    contents=messages[0]["content"],  # Use index [0] to get the first dictionary and extract the raw string via ["content"]
    config={
        'system_instruction': "You are a snarky assistant"
    }  # Use config with system_instruction to set the system role separate from the user query
)  # Call the generate_content method under the client's models interface
 
response.text  # Access the generated text directly using the simple text attribute of the response object

# HOW THIS CODE WORKS:
# 1. client = genai.Client() uses the modern standard google-genai library to initialize the model interface.
# 2. messages is a list of dictionaries; [0] accesses the first dictionary and ["content"] gets the string "What is 2 + 2?".
# 3. model="gemini-2.5-flash" specifies the lightweight, fast, and free-tier-friendly multimodal model.
# 4. contents accepts the query string directly; config holds system_instruction to set the AI's persona separately from the query.
# 5. response.text directly extracts the generated text string from the response object without deep nested dictionaries.
````

## And now let's build useful messages for gemini-2.5-flash, using a function


````python
# See how this function creates a Gemini-compatible structure

def messages_for(website):
    return [
        {"role": "user", "parts": [{"text": user_prompt_prefix + website}]}
    ]  # Wrap the query string in a dictionary with a 'text' key inside the parts list

# The previous code passed a raw string directly inside the 'parts' list (e.g., ["string"]).
# This was wrong because the google-genai library's Pydantic schema expects 'parts' to contain valid Part objects or dictionaries.
# The new code changed this by wrapping the text query inside a dictionary with a 'text' key (e.g., [{"text": "string"}]).
# This matches the expected schema of the Part object, allowing local validation to pass and the API call to proceed.
````

````python
# Try this out, and then try for a few more websites

messages_for(ed)
````

## Time to bring it together - the API for Gemini is very simple!


````python
# And now: call the Gemini API!

def summarize(url):
    website = fetch_website_contents(url)
    response = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=messages_for(website),
        config={
            'system_instruction': system_prompt
        }
    )
    return response.text
````

````python
summarize("https://edwarddonner.com")
````

````python
# A function to display this nicely in the output, using markdown

def display_summary(url):
    summary = summarize(url)
    display(Markdown(summary))
````

````python
display_summary("https://edwarddonner.com")
````
